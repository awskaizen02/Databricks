```python
#Importing types and defining a schema before reading in the csv

from pyspark.sql.types import *

hs_schema = StructType([
            StructField('country', StringType(), False), 
            StructField('indicator', StringType(), True), 
            StructField('date', StringType(), False), 
            StructField('year_week', StringType(), True), 
            StructField('value', StringType(), False), 
            StructField('source', StringType(), False), 
            StructField('url', StringType(), True)
])
```

```python
#reading in the csv file and passing the schema defined in the previous cell
hs_df = spark.read.csv("/Volumes/project03databricks/sample_ws/demovolume/hospital_admissions.csv",header = True, schema=hs_schema)
display(hs_df)
```

```python
#alternative syntax for passing in the schema inside the options methoid

hs_df1 = spark.read.options(header=True).schema(hs_schema).csv("/Volumes/project03databricks/sample_ws/demovolume/hospital_admissions.csv")
display(hs_df1)
```
