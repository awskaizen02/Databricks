# Optimize Command

```python
%sql
describe history employ
```

```python
%sql
optimize employ
zorder by id

-- this command is used to improve query performance on a delta table in databricks
```

OPTIMIZE

OPTIMIZE is a databricks Delta lake command. it compacts small files into larger, more efficient files. This is important  beacuse.

Databricks tables often get many small files from streaming, bathing jobs, or frequent update

Too many small files slow down queries and increate storage overhead

OPTIMIZE helps reduce the numbe of files by merging them into optimal sizedefault 1gb per file)

ZORDER BY id
ZORDERing is an indexing tecnique used in databricks delta table.

it co-lcocates related infomation in the same set of files to improve data skipping and query permance

ZORDER by id means that the table is optimized so that rows similar id values are stored physicall close together

without ZORDER:
  Databricks might scan many files to find the matching rows

with ZORDER:
  Databricks can skip unncessary files  beacuse related data is clusted together, leading to much faster queries

before optimize suppose this data is spread across many small parquet files  like : each file might contain random rows with mixed id values

employ/
  part-0001.snappu.parquet
  part-0002.snappu.parquet
  part-0003.snappu.parquet
  part-0004.snappu.parquet

Run optimize with ZOrder what happen: compaction: small files are erged int large files
Zopred cluster: rows with similar id values are store close together inside these large files.

employ/
  optimized-0001.snappy.parquet
  optimized-0002.snappy.parquet

After optimize : select * from employ where id = 3

quickly skips irrelevant files.
reads only the file that contains id = 3
reduces scan time, leading to fster query execution


OPTIMIZE  # REDUCES SMALL FILE OVERHEAD , IMPROVES READ PERFORMANCE
ZORDER    # IMPROVES DATA SKIPPING, DRASTICALLY REDUCING THE AMOUNT OF DATA SCCANED DURING FILTERE QUERIES

WITHOUR ZORDER : QUERY SCANS 100GB OF DATA
WITH ZORDER    : QUERY MAY SCAN ONLY 5GB OF DATA
